## 2. Importar la biblioteca


```python
# Importa la interfaz pública de la biblioteca y las masas nucleares de referencia.
from bohr_atom import (
    BohrAtom,
    PROTON_MASS_KG,
    HELIUM4_NUCLEUS_MASS_KG,
)

print("Biblioteca importada correctamente")
```

## 3. Crear un átomo de hidrógeno

El hidrógeno tiene número atómico `Z=1`. Utilizamos la masa del protón como masa del núcleo en la aproximación de hidrógeno-1.


```python
# Construye un objeto que representa el átomo de hidrógeno.
hydrogen = BohrAtom(
    Z=1,
    nucleus_mass_kg=PROTON_MASS_KG,
    name="Hidrógeno"
)

hydrogen
```

## 4. Consultar los parámetros físicos

La instancia almacena el número atómico, las masas, las cargas y la masa reducida del sistema electrón-núcleo.


```python
# Inspecciona los parámetros físicos y la masa reducida almacenados en el objeto.
print(f"Nombre: {hydrogen.name}")
print(f"Número atómico Z: {hydrogen.Z}")
print(f"Masa del electrón: {hydrogen.electron_mass_kg:.6e} kg")
print(f"Masa del núcleo: {hydrogen.nucleus_mass_kg:.6e} kg")
print(f"Masa reducida: {hydrogen.reduced_mass_kg:.6e} kg")
print(f"Carga del núcleo: {hydrogen.nuclear_charge_c:.6e} C")
print(f"Carga del electrón: {hydrogen.electron_charge_c:.6e} C")
```

## 5. Calcular energía y radio

El método `energy` devuelve la energía del nivel `n`. Admite julios (`"J"`) o electronvoltios (`"eV"`). El método `radius` admite metros, nanómetros, picómetros y ángstroms.


```python
# Calcula la energía y el radio del estado fundamental en distintas unidades.
n = 1

energy_ev = hydrogen.energy(n, unit="eV")
energy_j = hydrogen.energy(n, unit="J")
radius_pm = hydrogen.radius(n, unit="pm")

print(f"Energía para n={n}: {energy_ev:.6f} eV")
print(f"Energía para n={n}: {energy_j:.6e} J")
print(f"Radio para n={n}: {radius_pm:.6f} pm")
```

## 6. Crear un ion hidrogenoide: He⁺

El ion He⁺ tiene un único electrón y un núcleo con carga `+2e`, por tanto `Z=2`. Utilizamos la masa del núcleo de helio-4.


```python
# Construye He+ y compara la energía fundamental con la del hidrógeno.
helium_ion = BohrAtom(
    Z=2,
    nucleus_mass_kg=HELIUM4_NUCLEUS_MASS_KG,
    name="Helio ionizado (He⁺)"
)

print(helium_ion)
print(f"Energía fundamental de H: {hydrogen.energy(1):.6f} eV")
print(f"Energía fundamental de He⁺: {helium_ion.energy(1):.6f} eV")
```

## 7. Calcular una transición

La transición `n=3 → n=2` es una transición de emisión. El método devuelve las propiedades del fotón, incluyendo su energía, frecuencia y longitud de onda.


```python
# Evalúa las propiedades del fotón para la transición descendente n=3 hacia n=2.
transition = hydrogen.transition(3, 2)

for key, value in transition.items():
    print(f"{key}: {value}")
```
