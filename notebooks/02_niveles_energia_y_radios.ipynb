# 02. Niveles de energía y radios orbitales

Este cuaderno estudia la dependencia de la energía y del radio orbital con el número cuántico principal `n`.

## Objetivos

- Generar tablas de niveles energéticos.
- Comparar los niveles del hidrógeno y He⁺.
- Comprobar las dependencias `Eₙ ∝ -1/n²` y `rₙ ∝ n²`.
- Representar los resultados gráficamente.


> **Convención de documentación:** cada celda de código comienza con un comentario en español que explica su propósito. Las ecuaciones y los supuestos físicos se explican en las celdas Markdown; la implementación de los cálculos permanece en el paquete `bohr_atom`, no se duplica en este notebook.


```python
# Importa Matplotlib y crea los sistemas hidrogenoides que se compararán.
import matplotlib.pyplot as plt

from bohr_atom import (
    BohrAtom,
    PROTON_MASS_KG,
    HELIUM4_NUCLEUS_MASS_KG,
)

hydrogen = BohrAtom(
    Z=1,
    nucleus_mass_kg=PROTON_MASS_KG,
    name="Hidrógeno"
)

helium_ion = BohrAtom(
    Z=2,
    nucleus_mass_kg=HELIUM4_NUCLEUS_MASS_KG,
    name="He⁺"
)
```

## 1. Obtener una tabla de niveles

`energy_levels(n_max)` devuelve una lista de diccionarios. Cada elemento contiene el número cuántico, la energía y el radio expresado en picómetros.


```python
# Genera los primeros niveles energéticos y sus radios, y presenta los resultados en una tabla de texto.
levels = hydrogen.energy_levels(n_max=8, energy_unit="eV")

print(f"{'n':>3} {'Energía (eV)':>18} {'Radio (pm)':>18}")
print("-" * 43)

for level in levels:
    print(
        f"{level['n']:>3} "
        f"{level['energy']:>18.6f} "
        f"{level['radius_pm']:>18.6f}"
    )
```

## 2. Comprobar la dependencia con `n`

Para un sistema hidrogenoide, manteniendo fijos `Z` y la masa reducida:

\[
E_n = rac{E_1}{n^2}, \qquad r_n = r_1 n^2.
\]

La primera relación incluye el signo negativo de las energías ligadas: `E₁` es negativo y `Eₙ` se aproxima a cero cuando aumenta `n`.


```python
# Comprueba numéricamente las dependencias E_n proporcional a 1/n² y r_n proporcional a n².
print("Comprobación de la energía:")
E1 = hydrogen.energy(1, unit="eV")

for n in range(1, 6):
    En = hydrogen.energy(n, unit="eV")
    print(f"n={n}: E_n={En: .6f} eV | E_n/E_1={En/E1:.6f} | 1/n²={1/n**2:.6f}")

print("\nComprobación del radio:")
r1 = hydrogen.radius(1, unit="pm")

for n in range(1, 6):
    rn = hydrogen.radius(n, unit="pm")
    print(f"n={n}: r_n={rn:.6f} pm | r_n/r_1={rn/r1:.6f} | n²={n**2}")
```

## 3. Graficar energía en función de `n`


```python
# Representa la energía de los niveles del hidrógeno en función del número cuántico principal.
n_values = [level["n"] for level in levels]
energies = [level["energy"] for level in levels]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(n_values, energies, marker="o")
ax.axhline(0, linewidth=1, linestyle="--")
ax.set_xlabel("Número cuántico principal, n")
ax.set_ylabel("Energía (eV)")
ax.set_title("Niveles de energía del hidrógeno")
ax.set_xticks(n_values)
ax.grid(alpha=0.3)
plt.show()
```

## 4. Diagrama de niveles energéticos

El método `plot_energy_levels` representa los niveles permitidos y, por defecto, las transiciones descendentes entre los niveles dibujados.


```python
# Utiliza el método de la biblioteca para dibujar los niveles de energía.
hydrogen.plot_energy_levels(n_max=6)
plt.show()
```

## 5. Comparar hidrógeno y He⁺

La energía depende de `Z²`, mientras que el radio depende de `1/Z`, además de la pequeña corrección que introduce la masa reducida de cada sistema.


```python
# Calcula y compara las energías de H y He+ para los mismos números cuánticos.
n_values = list(range(1, 6))

hydrogen_energies = [hydrogen.energy(n) for n in n_values]
helium_energies = [helium_ion.energy(n) for n in n_values]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(n_values, hydrogen_energies, marker="o", label="H")
ax.plot(n_values, helium_energies, marker="s", label="He⁺")
ax.axhline(0, linewidth=1, linestyle="--")
ax.set_xlabel("Número cuántico principal, n")
ax.set_ylabel("Energía (eV)")
ax.set_title("Comparación de niveles energéticos")
ax.set_xticks(n_values)
ax.grid(alpha=0.3)
ax.legend()
plt.show()
```

## 6. Comparar radios orbitales


```python
# Compara los radios relativos electrón-núcleo de ambos sistemas.
hydrogen_radii = [hydrogen.radius(n, unit="pm") for n in n_values]
helium_radii = [helium_ion.radius(n, unit="pm") for n in n_values]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(n_values, hydrogen_radii, marker="o", label="H")
ax.plot(n_values, helium_radii, marker="s", label="He⁺")
ax.set_xlabel("Número cuántico principal, n")
ax.set_ylabel("Radio electrón-núcleo (pm)")
ax.set_title("Radios permitidos del modelo de Bohr")
ax.set_xticks(n_values)
ax.grid(alpha=0.3)
ax.legend()
plt.show()
```
