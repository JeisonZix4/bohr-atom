# 04. Visualización de niveles de energía y órbitas

Este cuaderno reúne los métodos de visualización de `BohrAtom`.

## Objetivos

- Dibujar diagramas de niveles de energía.
- Seleccionar las transiciones que se desean representar.
- Dibujar las órbitas circulares en el plano.
- Comparar cualitativamente las escalas orbitales de H y He⁺.

> Las órbitas circulares son una representación del modelo histórico de Bohr. No son trayectorias literales del electrón en la mecánica cuántica moderna.


> **Convención de documentación:** cada celda de código comienza con un comentario en español que explica su propósito. Las ecuaciones y los supuestos físicos se explican en las celdas Markdown; la implementación de los cálculos permanece en el paquete `bohr_atom`, no se duplica en este notebook.


```python
# Importa Matplotlib y construye hidrógeno y He+ para las comparaciones gráficas.
import matplotlib.pyplot as plt

from bohr_atom import (
    BohrAtom,
    PROTON_MASS_KG,
    HELIUM4_NUCLEUS_MASS_KG,
)

hydrogen = BohrAtom(
    Z=1,
    nucleus_mass_kg=PROTON_MASS_KG,
    name="Hidrógeno"
)

helium_ion = BohrAtom(
    Z=2,
    nucleus_mass_kg=HELIUM4_NUCLEUS_MASS_KG,
    name="He⁺"
)
```

## 1. Diagrama con todas las transiciones descendentes

Si no se especifica el argumento `transitions`, el método `plot_energy_levels` representa las transiciones descendentes entre los niveles incluidos.


```python
# Dibuja los niveles energéticos del hidrógeno.
hydrogen.plot_energy_levels(n_max=5)
plt.show()
```

## 2. Seleccionar transiciones concretas

Cada transición se especifica como una pareja `(n_inicial, n_final)`. Por ejemplo, las transiciones hacia `n=2` corresponden a líneas de la serie de Balmer.


```python
# Selecciona transiciones de la serie de Balmer que terminan en n=2.
balmer_transitions = [
    (3, 2),
    (4, 2),
    (5, 2),
]

hydrogen.plot_energy_levels(
    n_max=5,
    transitions=balmer_transitions
)
plt.show()
```

## 3. Dibujar órbitas en el plano

El método `plot_orbits` recibe una lista con los niveles que se quieren representar. Los radios se convierten a picómetros para mostrar una escala apropiada para estos sistemas.


```python
# Representa las órbitas circulares permitidas para varios niveles.
hydrogen.plot_orbits(levels=[1, 2, 3, 4])
plt.show()
```

## 4. Comparar las órbitas de H y He⁺

Como `BohrAtom` calcula radios para cada sistema usando su número atómico y masa reducida, podemos comparar sus radios para los mismos valores de `n`.


```python
# Compara las escalas orbitales de H y He+ en dos paneles.
levels = [1, 2, 3, 4]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

hydrogen.plot_orbits(levels=levels, ax=axes[0])
helium_ion.plot_orbits(levels=levels, ax=axes[1])

fig.tight_layout()
plt.show()
```

## 5. Comparar numéricamente los radios

La tabla siguiente muestra los radios electrón-núcleo para ambos sistemas. Recuerda que se trata de la coordenada relativa del problema de dos cuerpos.


```python
# Calcula y compara los radios orbitales en picómetros.
print(
    f"{'n':>3} {'H (pm)':>18} {'He⁺ (pm)':>18} "
    f"{'r(H)/r(He⁺)':>18}"
)
print("-" * 61)

for n in levels:
    r_h = hydrogen.radius(n, unit="pm")
    r_he = helium_ion.radius(n, unit="pm")
    print(f"{n:>3} {r_h:>18.6f} {r_he:>18.6f} {r_h/r_he:>18.6f}")
```

## 6. Interpretación

- Para un mismo número cuántico, el radio de He⁺ es menor que el de H.
- El aumento de `n` incrementa el radio proporcionalmente a `n²`.
- Los diagramas de energía y las órbitas son representaciones diferentes: uno muestra valores energéticos y el otro las separaciones permitidas.
- Ninguna de estas figuras representa una función de onda o una densidad de probabilidad.
